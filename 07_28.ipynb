```python
# break 실습 2
N = int(input(" 최대 횟수를 입력해주세요: "))

for i in range(N):
    n = int(input("측정값을 입력해주세요: "))
    if n > 80:
        print("이상값 발견, 중단")
        break

    print("정상")   
```

```python
# continue 실습 3

sum = 0

N = int(input("횟수를 입력하세요: "))

for i in range(N):
    n = int(input("측정값을 입력하세요: "))

    if n < 0:
        continue
    sum += n

print(sum)
```

```python
# 최댓값 최솟값 찾기, 실습4

N = int(input("시작값을 입력해주세요: "))
K = int(input("입력횟수를 지정해주세요: "))
max_value = N
min_value = N

for i in range(K):
    m = int(input("값을 입력해 주세요: "))
    if m > N:
        max_value = m
    if m < N:
        min_value = m

print(f"최댓값: {max_value}, 최솟값: {min_value}")
```

```python
# 실습5. 플래그로 조건 만족 값 검색하기
found = False

N = int(input("횟수를 입력하세요: "))

for i in range(N):
    k = int(input("측정값을 입력해주세요: "))

    if k > 80:
        found = True
        break

if found:
    print("발견")

else:
    print("없음")
```

```python
#실습6. 종합 미니 프로그램 만들기

N=5
total = 0
over_count = 0
found = False
for i in range(5):
    n = int(input("측정값을 입력해주세요: "))
    total += n

    if n > 80:
        over_count += 1
        found = True

print(f"평균: {total / N}, 초과 개수: {over_count}, 이상 여부: {found}")

```

```python
# 실습 6: 센서 데이터 종합 분석하기

temps = [27, 31, 41, 19, 65, 37, 18]
count = 0
total = 0
em = []

for i in temps:
    total += i
    count += 1
    if i > 30:
        em.append(i)

new_count = 0
new_total = 0
for n in em:
    new_count += 1
    new_total += n

print(f"전체 평균: {total / count}, 고온 개수: {new_count}, 고온 평균: {new_total / new_count}")
```

```python
sentence = "the quick fox the lazy dog the end"
# 여기에 코드 작성
words = sentence.split(" ")
print("단어",len(words),"개")
print(f'the의 등장 횟수: {words.count("the")}')

long_words = words[0]
for i in range(1, len(words)):
    if len(words[i-1]) > len(long_words):
        long_words = words[i-1]

print("가장 긴 단어:", long_words)
```

```python
print("[15]")
sensors = [["펌프", 72], ["모터", 91], ["팬", 65], ["압축기", 88]]
# 여기에 코드 작성
total_temp = 0
total_count = 0
warning = []

for i in range(len(sensors)):
    total_temp += sensors[i][1]
    total_count += 1

    if sensors[i][1] > 80:
        warning.append(sensors[i][0])
avg_temp = total_temp / total_count

print(
    f'{"="*5}관제리포트{"="*5}\n 감시 설비: {total_count}대\n 평균 온도: {avg_temp}도\n 경고 설비: {len(warning)} - {warning}'
)

print()
print("총정리 끝! 여기까지 스스로 풀었다면 기초 과정 완주입니다")
```
