```python
#함수, 매개변수 연습문제 6
def sensor_summary(a, b):
    if a > b:
        max_value = a
        min_value = b
    else:
        max_value = b
        min_value = a
    return max_value, min_value, (a+b) / 2

lowest, highest, average = sensor_summary(78,92)
print("최저:", lowest)
print("최고:", highest)
print("평군:", average)


```

```python
#실습4. 센서 분석 함수 세트 만들기

def average(values):
    return sum(values) / len(values)

def status(avg, limit = 90):
    if avg > limit:
        print("경고")
    else:
        print("정상")

avg = average([78, 92, 85])

print(avg)
status(avg)
```
