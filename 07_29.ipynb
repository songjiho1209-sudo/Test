```python
yesterday = [("S01", "정상"), ("S02", "경고"), ("S03", "경고")]
today = [("S01", "정상"), ("S02", "경고"), ("S03", "정상"), ("S04", "경고")]

y_warn = set()
t_warn = set()

for sensor, status in yesterday:
    if status == "경고":
        y_warn.add(sensor)

for sensor, status in today:
    if status == "경고":
        t_warn.add(sensor)


print("신규 경고: ", t_warn.difference(y_warn) )
print("지속 경고: ", t_warn.intersection(y_warn))
```

```python
orders = [("A01", "연필"), ("A02", "공책"), ("A03", "연필"), ("A04", "지우개"), ("A05", "공책")]
regi = {"연필", "공책", "지우개"}

product = set()
sen = []

for num, name in orders:
    product.add(name)

d=set()

for num, name in orders:
    sen.append(name)

for name in sen:
    if sen.count(name) >= 2:
        d.add(name)

print(sorted(product))
print(sorted(d))
print("미등록 상품: ", sorted(product.difference(regi)))
```
