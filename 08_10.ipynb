```python
#실습3
pressure_cyl = np.loadtxt('data/11_diecasting_series.csv',
                  delimiter = ",",
                  skiprows =1,
                  usecols = 2
                )

pressure_cyl[0] = 9999

sum__err_pressure_cyl = pressure_cyl.sum()

avg__err_pressure_cyl = sum__err_pressure_cyl / len(pressure_cyl)

pressure_cyl_adj = np.clip(pressure_cyl, 100, 300)

avg_pressure_cly_adj = pressure_cyl_adj.sum() / len(pressure_cyl)

print(f"오류가 있을때 평균: {avg__err_pressure_cyl}")
print(f"조정 후 평균: {avg_pressure_cly_adj}")
```

```python
# np.cumsum(변수)
# 누적합을 구해줌
# 지금까지는 반복문을 돌면서 변수에 값을 더해 직접 누적합을 구했으나
# comsum을 사용하면 누적합을 바로 확인할 수 잇음
# cumsum은 배열의 전체 누적값을 반환하는 것이 아니라
# 기존 배열과 동일한 길이라 각 값을 하나의 누적합 배열로 변환

new_arr = np.array([1, 2, 3, 4, 5])
print(np.cumsum(new_arr))
```

```python
import numpy as np

data = np.loadtxt(
    'data/11_diecasting_series.csv',
    delimiter=',',
    skiprows=1,
    usecols=(0, 1, 2)
)

time = data[:, 0]
ct = data[:, 1]
pressure_cyl = data[:, 2]


idx_max_ct = np.argmax(ct)

time_max_ct = time[idx_max_ct]
max_ct = ct[idx_max_ct]

print(f'사이클타임이 가장 컸던 시점: {time_max_ct}, 사이클타임 최대값: {max_ct}')


change_ct = np.abs(np.diff(ct))
idx_max_change_ct = np.argmax(change_ct)

print(
    f'사이클타임 변화량이 가장 컸던 구간: '
    f'시점 {time[idx_max_change_ct]} -> {time[idx_max_change_ct + 1]}, '
    f'변화량: {change_ct[idx_max_change_ct]}'
)

warning = (ct >= 50) | (pressure_cyl < 150)

warning_count = warning.sum()
warning_ratio = warning_count / len(warning)

print(f'경고 개수: {warning_count}')
print(f'경고 비율: {warning_ratio}')


top3_idx = np.argsort(ct)[::-1][:3]

print('사이클타임 상위 3개:')
for idx in top3_idx:
    print(f'시점: {time[idx]}')
```
