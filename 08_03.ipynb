```python
#실습2. 설비명만 셋으로 만들기
equipment_name = set()
with open('data/08_press.csv', 'r', encoding='utf-8') as f:
    for line in f:
        equipment_id = line.strip().split(",")[0]

        if equipment_id=="설비ID":
            continue

        equipment_name.add(equipment_id)


eqipment_list = sorted(list(equipment_name))
print(eqipment_list)
        
```

```python
#실습4. 설비명만 셋으로 만들기 with csv
import csv, os
equipment_name = set()

with open('data/08_press.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(f)
    for row in reader:
        equipment_name.add(row[0])


eqipment_list = sorted(list(equipment_name))
print(eqipment_list)
        
```
