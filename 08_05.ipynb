```python
# narray: 배열 (우리가 학습했던 리스트와 다른 자료형)
# numpy에서 자료를 담는 그릇
# 같은 자료형만 담음 (ex. 정수 배열, 실수 배열)
# 정해진 형태를 가지고 잇음 (한 줄로 구성되어 있거나 표 모양)

# 리스트로 배열의 요소들에 *5릃 하는 코드 작성

tiems=0
type_list= [1, 2, 3, 4, 5]
times_5 = []
for num in type_list:
    times = num*5
    times_5.append(times)

print(times_5)

print(times_5 +[5])
```

```python
temp = []
for value in body:
    temp.append(float(value[2]))
temp = np.array(temp)
F_temp = 1.8*temp+32

print(F_temp)
```

```python
# for문을 짧게 작성하는 방법
# 리스트 컴프리헨션
# 리스트에_삽입할_값 for 변수 in 리스트명
# ex)) float(row[2]) for row in body

temp = np.array([float(value[2]) for value in body])
# 대괄호로 감싸주면 for 문을 돌면서 뽑아낸 온도값이
# 밖에 있는 대괄호 안에 하나씩 자동으로 들어감

F_temp = temp*1.8+32

print(F_temp)
```

```python
#ndim: 배열의 차원을 출력해줌 = 리스트 안의 리스트의 개수
# 직접 리스트를 입력한 배열의 경우 보기 좋은
# but 변수로 배열을 만드는 경우
# 해당 변수에 담긴 값을 확인하기는 번거로움
# 배열로 바꾸고자 하는 리스트의 길이가 매우 길 경우 확인이 불편함
# 그럴 때 간단하게 배열의 차원을 확인하기 위해서 사용

# print(a.ndim)
# print(b.ndim)

# ndim은 속성이기 때문에 괄호를 사용하지 않음
# 1차원은 값이 왼쪽에서 오른쪽으로 나열되어있는 형태
# 2차원은 표 형태로 행과 열이 있는 형태로설비 데이터는 보통 행과 열로 이루어져있고,
# 행= 시점, 열 = 센서값을 가진 2차원의 표 형태

```
