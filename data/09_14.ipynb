```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('25_mimii_features_94_260914_2.csv')

normal = df[df['label'] == 0]
soon = df[df['label'] == 1]

plt.scatter(
    normal['rms'],
    normal['spectral_centroid'],
    color='blue',
    label='정상 (0)'
)

plt.scatter(
    soon['rms'],
    soon['spectral_centroid'],
    color='red',
    label='임박 (1)'
)

plt.xlabel('rms')
plt.ylabel('spectral_centroid')
plt.title('RMS와 Spectral Centroid 산점도')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

normal = df[df['label'] == 0]
soon = df[df['label'] == 1]

plt.scatter(normal['rms'], normal['zero_crossing_rate'],
            color='blue', label='정상 (0)')

plt.scatter(soon['rms'], soon['zero_crossing_rate'],
            color='red', label='임박 (1)')

plt.xlabel('rms')
plt.ylabel('zero_crossing_rate')
plt.title('RMS와 Zero Crossing Rate 산점도')
plt.legend()
plt.grid(True)
plt.show()
```
