```python
df = pd.read_csv('설비센서_샘플.csv')

a = df[df['설비명'] == 'A호기']

plt.plot(a['일자'], a['온도'])
plt.show()
plt.plot(a['일자'], a['온도'], color='green')
plt.show()
plt.plot(a['일자'], a['온도'],color='red', linestyle = '--', marker='o')
plt.show()
```

```python
c = df[df['설비명'] == 'C호기']
plt.figure(figsize=(10,4))
plt.plot(c['일자'], c['온도'], marker='o', label = 'C호기')
plt.plot(a['일자'], a['온도'], label= 'A호기')
plt.legend(loc='upper left')
plt.ylabel('온도(도)')
plt.xlabel('일자')
plt.title('온도변화')
plt.xticks([1,2,3,4,5], ['월','화','수','목','금'])
plt.grid(True, alpha = 1.0)

plt.show()
```

```python
c = df[df['설비명'] == 'C호기']
plt.figure(figsize=(10,4))
plt.plot(c['일자'], c['온도'], label = 'C호기', color = 'red', marker = 's')
plt.plot(a['일자'], a['온도'], label= 'A호기', color = 'blue', marker = 'o')
plt.title('A호기 vs C호기 온도 변화')
plt.xlabel('일자')
plt.ylabel('온도(도)')
plt.legend(loc='upper left')
plt.show()
```

```python

plt.figure(figsize=(10, 4))

for name, color in [("A호기", "blue"), ("B호기", "green"), ("C호기", "red")]:
    sub = df[df["설비명"] == name]

    plt.plot(
        sub["일자"],
        sub["온도"],
        label=name,
        color=color,
        marker="o"
    )

plt.title("설비별 온도 비교")
plt.xlabel("일자")
plt.ylabel("온도(℃)")
plt.legend()

plt.tight_layout()
plt.savefig("설비별_온도_비교.png")
plt.show()
```

```python

df = pd.read_csv("설비센서_샘플.csv")

avg = df.groupby("설비명")["온도"].mean()

plt.bar(
    avg.index,
    avg.values,
    color=["blue", "blue", "red"],
    width=0.5)

plt.title("설비별 평균 온도")
plt.ylabel("온도(℃)")
plt.show()

plt.barh(
    avg.index,
    avg.values,
    color=["blue", "blue", "red"],
    height=0.5)

plt.title("설비별 평균 온도")
plt.xlabel("온도(℃)")
plt.show()
```

```python
state = df["상태"].value_counts()

plt.bar(state.index, state.values)

plt.title("상태별 데이터 개수")
plt.ylabel("개수")
plt.show()

zone = df.groupby("구역")["진동"].mean()

plt.bar(zone.index, zone.values)

plt.title("구역별 평균 진동")
plt.ylabel("진동(mm/s)")
plt.show()
```

```python
plt.hist(df["온도"])
plt.title("온도 분포")
plt.xlabel("온도(℃)")
plt.ylabel("개수")
plt.show()

plt.hist(df["온도"], bins=5)
plt.title("온도 분포")
plt.xlabel("온도(℃)")
plt.ylabel("개수")
plt.show()

plt.hist(df["온도"], bins=20)
plt.title("온도 분포")
plt.xlabel("온도(℃)")
plt.ylabel("개수")
plt.show()
```

```python
plt.hist(df["온도"], bins=8)

mean_temp = df["온도"].mean()
std_temp = df["온도"].std()

print("평균 온도:", round(mean_temp, 1), "℃")
print("표준편차:", round(std_temp, 1), "℃")

high_temp = df[df["온도"] >= 78]

print("78℃ 이상 데이터:", len(high_temp), "개")

plt.title("온도 분포")
plt.xlabel("온도(℃)")
plt.ylabel("개수")
plt.show()
```

```python
color = df["상태"].map({
    "정상": "blue",
    "점검": "red"})

plt.scatter(
    df["온도"],
    df["진동"],
    color=color,
    alpha=0.6)

plt.axhline(
    y=4.0,
    color="red",
    linestyle="--")

plt.title("온도와 진동의 관계")
plt.xlabel("온도(℃)")
plt.ylabel("진동(mm/s)")
plt.show()
```
