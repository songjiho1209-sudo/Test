```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
plt.rcParams["font.family"] = "Malgun Gothic"
df = pd.read_csv("18_equipment_sensor.csv")
sns.histplot(data=df, x='전류')

plt.title('전류 분포')
plt.xlabel('전류(A)')
plt.ylabel('개수')

plt.show()
```

```python
for b in [10, 20, 40]:
    sns.histplot(data=df, x='전류', bins=b, kde=True)

    plt.title(f'전류 분포 bins={b}')
    plt.xlabel('전류(A)')
    plt.ylabel('개수')

    plt.show()
```

```python
for col, unit in [('온도', '°C'), ('진동', 'mm/s'), ('전류', 'A')]:
    sns.histplot(data=df, x=col, bins=20, kde=True)

    plt.title(f'{col} 분포')
    plt.xlabel(f'{col}({unit})')
    plt.ylabel('개수')

    plt.show()
```

```python
sns.boxplot(data=df, y='전류')

plt.title('전류 박스플롯')
plt.ylabel('전류(A)')

plt.show()
```

```python
sns.boxplot(data=df, x='설비라인', y='온도')
plt.title('라인별 온도 분포')
plt.xlabel('설비라인')
plt.ylabel('온도')
plt.show()

sns.boxplot(data=df, x='설비라인', y='온도', hue='판정')
plt.title('라인별·판정별 온도 비교')
plt.xlabel('설비라인')
plt.ylabel('온도')
plt.show()
```

```python

sns.countplot(data=df, x='설비라인', hue='설비라인', palette='Set2', legend=False)
plt.title('라인별 데이터 개수')
plt.xlabel('설비라인')
plt.ylabel('개수')
plt.show()

sns.barplot(data=df, x='설비라인', y='전류', hue='설비라인', palette='pastel', legend=False)
plt.title('라인별 평균 전류')
plt.xlabel('설비라인')
plt.ylabel('평균 전류(A)')
plt.show()
```

```python
sns.countplot(data=df, x='설비라인', hue='판정')
plt.title('라인별 양품·불량 개수')
plt.xlabel('설비라인')
plt.ylabel('개수')
plt.show()

df.groupby(['설비라인', '판정']).size().unstack(fill_value=0)
```

```python
sns.scatterplot(data=df, x='온도', y='압력')

plt.title('온도와 압력의 관계')
plt.xlabel('온도(°C)')
plt.ylabel('압력')
plt.show()
```
