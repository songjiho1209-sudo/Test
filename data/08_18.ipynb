```python
dfs = pd.read_csv('15_01_secom_교육샘플.csv')
cnt = dfs.isna().sum()

ratio = round((cnt/len(dfs)*100),1)

표 = pd.DataFrame({
    "결측치 개수": cnt,
    "결측 비율(%)": ratio
})

print(표)

print("결측치가 하나 이상 있는 컬럼")
print(표[표["결측치 개수"] > 0])

print("가장 결측치가 많은 컬럼")
print(표[표["결측치 개수"] == 표["결측치 개수"].max()])
```

```python
dfs = pd.read_csv("15_01_secom_교육샘플.csv")

cnt = dfs.isna().sum()
ratio = round((cnt / len(dfs) * 100), 1)

summary = pd.DataFrame({
    "개수": cnt,
    "비율": ratio
})

summary2 = summary[summary["개수"] > 0].copy()

def 처리방향(x):
    if x < 5:
        return "대체"
    elif x >= 40:
        return "제거 고려"
    else:
        return "검토"

summary2["처리방향"] = summary2["비율"].apply(처리방향)

summary2.to_csv("결측요약.csv", encoding="cp949")

summary2
```

```python
ratio_sorted = ratio.sort_values(ascending=False)

print("결측 비율 상위 3개 컬럼")
print(ratio_sorted.head(3))

행결측 = dfs.isna().sum(axis=1)

print("\n행별 결측치 개수 상위 10개")
print(행결측.sort_values(ascending=False).head(10))

print("\n결측치가 하나 이상 있는 행의 개수")
print((행결측 >= 1).sum())

print("\n결측치가 5개 이상 포함된 행과 컬럼 수")
print(dfs[행결측 >= 5].shape)
```

```python
df = pd.read_csv('15_01_secom_교육샘플.csv')
```

```python
비교 = pd.DataFrame({
    "방식": ["원본", "행삭제", "thresh20"],
    "행": [
        len(df),
        len(df.dropna()),
        len(df.dropna(thresh=20))
    ]
})

비교["손실률"] = round((1 - 비교["행"] / len(df)) * 100, 1)

print(비교)
```

```python
# 센서17 평균과 중앙값 계산
avg = df["센서17"].mean()
mid = df["센서17"].median()

print("센서17 평균:", round(avg, 2))
print("센서17 중앙값:", round(mid, 2))


fill_avg = df["센서17"].fillna(avg)


fill_mid = df["센서17"].fillna(mid)

# 대체 후 남아있는 결측치 개수 확인
print("fill_avg 결측치 개수:", fill_avg.isna().sum())
print("flii_mid 결측치 개수:", fill_mid.isna().sum())
```
