```python
Q1 = df2['실린더압력'].quantile(0.25)
Q3 = df2['실린더압력'].quantile(0.75)

IQR = Q3 - Q1
L = Q1 -1.5*IQR
U = Q3 + 1.5*IQR

m = (df2['실린더압력'] < L) | (df2['실린더압력'] > U)

masked = df2['실린더압력'].mask(m)

med = masked.median()
filled = masked.fillna(med)

removed = df2.loc[~m, '실린더압력']

clipped = df2['실린더압력'].clip(lower=L, upper=U)

mean_original = df2['실린더압력'].mean()
mean_removed = removed.mean()
mean_clipped = clipped.mean()
mean_filled = filled.mean()

print(f'평균은 약 {mean_original:.2f} → 제거 {mean_removed:.2f} / 보정 {mean_clipped:.2f} / 중앙값 채움 {mean_filled:.2f}')





```
