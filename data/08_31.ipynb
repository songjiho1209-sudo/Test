```python
df = pd.read_csv('20_cmapss_fd001_sample_65_260831.csv')
e1 = df[df['unit_nr'] ==  1].sort_values('time_cycles')

e1['ma5'] = e1['s_4'].rolling(window=5).mean()
e1['std5'] = e1['s_4'].rolling(window=5).std()
print(e1[['ma5', 'std5']].round(3).head())
```

```python
df = pd.read_csv('20_cmapss_fd001_sample_65_260831.csv')
e2 = df[df['unit_nr'] ==  2].sort_values('time_cycles').reset_index(drop=True)

e2['ma5'] = e2['s_11'].rolling(window=5).mean()
print(e2['ma5'].head(6).round(3))
```

```python
e2 = df[df['unit_nr'] ==  2].sort_values('time_cycles').reset_index(drop=True)

e2['std5'] = e2['s_11'].rolling(window=5).std()
print(e2['std5'].iloc[4].round(3))

plt.plot(e2["time_cycles"], e2["s_11"], alpha=0.4, label="s_11")
plt.plot(e2['time_cycles'], e2['std5'], linewidth=2, label = 'std5')
plt.legend()
plt.show()
```
