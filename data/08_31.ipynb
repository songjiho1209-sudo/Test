```python
import pandas as pd

df = pd.read_csv("20_cmapss_fd001_sample_65_260831.csv")

# 2번 엔진만 선택 후 작동 회차 순 정렬
df_2 = df[df["unit_nr"] == 2].sort_values("time_cycles")

# 첫 회차, 마지막 회차
first = df_2.iloc[0]
last = df_2.iloc[-1]

# 예: 상승 센서 s_2, 하강 센서 s_7 확인
print("s_11 첫 값:", first["s_11"])
print("s_11 마지막 값:", last["s_11"])

print("s_7 첫 값:", first["s_7"])
print("s_7 마지막 값:", last["s_7"])
```
