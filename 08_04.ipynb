```python
# try-except문
# try: 에러가 발생할 수 잇는 위험한 코드 작성
# except: 에러가 발생했을 때 이에 대응하는 코드를 작성

num = input("숫자를 입력하세요: ")
try:
    temp = int(num)
except ValueError:
    print("입력하신 값은 숫자가 아닙니다")
```

```python
#실습2
while True:
    temp_input = input("온도를 입력하세요: ")

    if temp_input.strip() == "":
        print("값을 입력하세요.")
        continue

    try:
        temp = float(temp_input)
    except ValueError:
        print("숫자로 변환할 수 없습니다.")
        continue

    print(f"온도: {temp}")
    print("프로그램 종료")
    break
```

```python
with open("data/09_ict_inspection_dirty.csv", "r", encoding="utf-8") as f:
    rows = f.readlines()
    total = 0.0
    skipped = 0
    for line in rows[1:]:                 # 헤더 제외
        cols = line.strip().split(",")
        try:
            total += float(cols[2])       # 측정값(2열)
        except (ValueError, IndexError):
            skipped += 1
            continue                      # 불량 줄은 건너뜀
print("합계:", round(total, 1), "건너뜀:", skipped)
```

```python
#실습. 부품별 csv 만들기
#부품종류는 E2, F2, B2, A2, D2
A2 = []
B2 = []
D2 = []
E2 = []
F2 = []
import csv
with open("data/09_ict_inspection.csv", "r", encoding='utf-8') as f:
    reader = csv.reader(f)

    header = next(reader)
    
    for row in reader:
        if row[1] == "A2":
            A2.append(row)
        elif row[1] == "B2":
            B2.append(row)
        elif row[1] == "D2":
            D2.append(row)
        elif row[1] == "E2":
            E2.append(row)
        elif row[1] == "F2":
            F2.append(row)

with open("data/A2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(A2)

with open("data/B2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(B2)

with open("data/D2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(D2)

with open("data/F2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(F2)

with open("data/E2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(E2)
```

```python
#실습2. 측정값이 하한치와 상한치 내에 있는 행만 
range_ok = []
import csv
with open("data/09_ict_inspection.csv", "r", encoding='utf-8') as f:
    reader = csv.reader(f)

    header = next(reader)
    
    for row in reader:
        if row[2] == "" or row[2] == "0.0":
            continue
        if float(row[5]) < float(row[2]) < float(row[4]):
            range_ok.append(row)

with open("data/range_ok.csv", "w", encoding='utf-8', newline="") as f:

    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(range_ok)
        
```

```python
import csv

def is_normal(measure, lower, upper):
    return measure != 0 and lower < measure < upper

range_ok = []

with open("data/09_ict_inspection.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)

    header = next(reader)

    for row in reader:
        if row[2] == "":
            continue

        measure = float(row[2])
        upper = float(row[4])
        lower = float(row[5])

        if is_normal(measure, lower, upper):
            range_ok.append(row)

is_normal(1,0,2)
```

```python
def check_value(value, lower, upper):
    val = float(value) #빈 값 or 문자면 valueError가 발생함
    if val == 0 or not (lower <= value <= upper):
        raise ValueError(f'부적절한 값 {val}')
    return val
    # 빈 값, 문자열, 측정값이 0 그리고 정상범위가 아니면 ValueError가 발생


```
