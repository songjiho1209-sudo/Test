```python
# 정규화: 값의 범위를 0~1사이로 정리하는 것
# 단위가 다른 값들 비교하기 위해 필요함
# (값 - 최솟값) / (최댓값 - 최솟값) : 정규화 식

arr_scores = np.array([61, 78, 59, 20, 83])

print(arr_scores >= 60)
```

```python
list_scores = [61, 78, 59, 20, 83]

for score in list_scores:
    print(score >= 60) 
```

```python
print(arr_scores[arr_scores >= 60])
# 조건을 만족하는 값만 남김
# 60을 넘는 값은 어떤 값인지를 알 수 있음
# 조건을 만족하지 못한다면 버려버림
# 불리언 인덱싱 결과는 기존 크기와 같거나 작아짐

```

```python
# 실습: 회전수의 평균, 중앙값, 분산, 표준편차 구하기

rpm = np.array(body)[:,4].astype(float)
avg_rpm = round(rpm.mean(), 1)
median_rpm = round((np.median(rpm)),1)
var_rpm = round(rpm.var(),1)
std_rpm = round(rpm.std(), 1)

print(f"회전수 평균: {avg_rpm}, 중앙값: {median_rpm}, 분산: {var_rpm}, 표준편차: {std_rpm}")
```

```python
# axis 개념 (row, column의 방향을 지정해줌.)
# axis=0은 위에서 아래로, 즉 열방향. 한개의 column의 값을 모음
# axis=1은 좌에서 우로(우에서 좌로), 즉 row 방향. 한 row의 값을 모음

import numpy as np
# 시험 점수
movies = np.array([["Runtime", "Relese_year", "Age_Rating"],
            [130, 2026, 13],
            [156, 2026, 15]
])

movies_body = movies[1:,].astype(int)

print("movies_body.mean():", movies_body.mean(axis=0))
# 결론적으로 평균 결과는 float 배열로 반환됨

# 행별 평균을 구하는 방법: axis = 1로 지정하면 됨
# 열별 평균을 구하는 방법: axis = 0으로 지정하면 됨
# 사실 current data set 기준으로는 행별 평균은 의미가 없음 because 단위가 다른 값이 섞여있다
# 특히 설비 데이터는 열마다 단위가 다르기 때문에 행별 평균을 내는 것은 의미가 없다. axis=0을 압도적으로 많이 사용

```

```python
#실습 1. 최대 실린더압력이 가장 큰 시점 찾기
datas = np.loadtxt('data/11_diecasting_series.csv',
                  delimiter = ",",
                  skiprows =1,
                  usecols = (0,1,2,3,4)
                )

ti = datas[:,0]
cyl = datas[:,2] 

idx_cylmax = np.argmax(cyl)
timax = ti[idx_cylmax]

print(f"가장 큰 값이 있는 인덱스: {idx_cylmax}, 실제 사이클번호: {timax}, 최대실린더압력: {cyl[idx_cylmax]}")
```

```python
ct = datas[:,1]
idx_ctmax = np.argmax(ct)
print(f'사이클타임 - 평균: {ct.mean()}, 중앙값: {np.median(ct)}, 최댓값: {ct.max()}, 최솟값: {ct.min()}, 표준편차: {ct.std()}, 사이클타임이 제일 긴 시점: {ti[idx_ctmax]} ')
print(f'실린더 압력 - 평균: {cyl.mean()}, , 중앙값: {np.median(cyl)}, 최댓값: {cyl.max()}, 최솟값: {cyl.min()}, 표준편차: {cyl.std()}, 사이클타임이 제일 긴 시점: {ti[idx_cylmax]}')
```
