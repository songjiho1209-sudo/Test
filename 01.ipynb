```python
# =============================================================================
#  1주차 종합 실습 ― 설비 점검 리포트 만들기  (01_02 ~ 03_06 전 범위)
# -----------------------------------------------------------------------------
#  지금까지 배운 모든 것을 하나로 엮어, 지저분한 센서 로그 한 줄을
#  사람이 읽는 "점검 리포트"로 바꾸는 프로그램을 단계별로 완성합니다.
#
#  · 미션은 서로 이어집니다. [1]부터 순서대로 푸세요.
#    (앞 미션에서 만든 변수를 뒤 미션에서 계속 사용합니다)
#  · "여기에 코드 작성" 아래에 직접 코드를 씁니다.
#  · 다 풀면 99_종합실습_정답.py 와 비교합니다.
#  · 온라인 연습장( https://ethan022.github.io/pdm_2th/ )에서 풀어도 됩니다.
#
#  ※ if · for · 함수 정의는 아직 안 배웠으므로 쓰지 않고 풀 수 있습니다.
#    리스트는 split 결과를 번호로 꺼내는 것까지만 사용합니다.
# =============================================================================


# =============================================================================
#  [1] 워밍업 ― 오류 3개 고치기                                     (01_02)
# -----------------------------------------------------------------------------
#  아래 세 줄에는 서로 다른 오류가 하나씩 있습니다.
#  주석(#)을 하나씩 풀어 오류 메시지를 읽고 → 종류를 파악하고 → 고쳐서
#  세 줄이 모두 정상 출력되게 만드세요.
#
#     print(설비 점검 시작)          ← 오류 1
#     print("가동률:", 87.5         ← 오류 2
#     print("온도: " + 91.5)        ← 오류 3
#
#  예상 결과:  설비 점검 시작  /  가동률: 87.5  /  온도: 91.5
# -----------------------------------------------------------------------------
print("[1]")
# 여기에 고친 코드 작성
print("설비 점검 시작")
print("가동률: ", 87.5)
print("온도: " + "91.5")

# =============================================================================
#  [2] 설비 기본 정보 ― 변수와 자료형                               (02_01 · 02_02)
# -----------------------------------------------------------------------------
#  이름에 어울리는 자료형으로 값을 저장하고, 값과 자료형을 함께 출력하세요.
#
#     equip_name    설비 이름   "PUMP-A-07"
#     equip_temp    온도        91.5
#     check_count   점검 횟수   3
#     is_running    가동 중     참
#
#  예상 결과 예)  PUMP-A-07 <class 'str'>
# -----------------------------------------------------------------------------
print("[2]")
# 여기에 코드 작성
equip_name = "PUMP-A-07"
equip_temp = 91.5
check_count = 3
is_running = "참"

print("설비 이름= ", equip_name, type(equip_name))
print("온도= ", equip_temp, type(equip_temp))
print("점검 횟수= ", check_count, type(check_count))
print("가동 중=", is_running, type(is_running))


# =============================================================================
#  [3] 가동 지표 계산 ― 연산자                                      (02_03)
# -----------------------------------------------------------------------------
#  주어진 값으로 세 가지를 계산해 출력하세요.
#
#  (1) 총 가동 1725분을  //, % 로  "가동 시간: 28 시간 45 분" 형태로
#  (2) 24시간 중 21시간 가동 → 가동률(%) 계산  (예상: 87.5)
#  (3) 전체 800개 중 불량 28개 → 불량률(%) 계산  (예상: 3.5)
#      ※ 3.5 뒤에 긴 오차가 붙어도 정상입니다 (02_02 실수 오차).
#        round(값, 1) 로 다듬어 한 번 더 출력해 보세요.
#
#  ※ (2)(3) 결과는 run_rate, defect_rate 변수에 저장해 두세요.
#    마지막 리포트에서 다시 사용합니다.
# -----------------------------------------------------------------------------
print("[3]")
total_min = 1725
run_h = 21
all_h = 24
total = 800
defect = 28
# 여기에 코드 작성
print("총 가동시간=", total_min // 60, "시간", total_min % 60, "분")
print("가동률=", run_h / all_h * 100)
print("불량률=", round(defect / total * 100, 1))
run_rate = run_h / all_h * 100
defect_rate = defect / total * 100

# =============================================================================
#  [4] 로그 한 줄 받기 ― 앞뒤 공백 정리                              (03_04)
# -----------------------------------------------------------------------------
#  센서에서 아래처럼 지저분한 로그 한 줄이 들어왔습니다.
#  (1) 정리 전 모습을 [ ] 로 감싸서 출력
#  (2) 앞뒤 공백을 제거해 log 변수에 저장하고, 다시 [ ] 로 감싸서 출력
#
#  예상 결과:
#     [  20260724, PUMP-A-07 , WARNING , temp=91.5  ]
#     [20260724, PUMP-A-07 , WARNING , temp=91.5]
# -----------------------------------------------------------------------------
print("[4]")
raw_log = "  20260724, PUMP-A-07 , WARNING , temp=91.5  "
# 여기에 코드 작성
print("[" + raw_log + "]")
log = raw_log.strip()
print("[" + log + "]")
# =============================================================================
#  [5] 쉼표로 나누고 조각 정리                                       (03_05 · 03_04)
# -----------------------------------------------------------------------------
#  (1) log 를 쉼표(,) 기준으로 나눠 리스트를 출력해 보세요.
#      → 조각들 앞뒤에 공백이 남아 있는 것을 눈으로 확인!
#  (2) 각 조각을 번호로 꺼내 strip() 으로 정리해 네 변수에 저장하세요.
#
#     date_text     0번 조각   →  "20260724"
#     code          1번 조각   →  "PUMP-A-07"
#     status_text   2번 조각   →  "WARNING"
#     measure_text  3번 조각   →  "temp=91.5"
#
#  (3) 네 변수를 각각 [ ] 로 감싸 출력해 공백이 사라졌는지 확인하세요.
# -----------------------------------------------------------------------------
print("[5]")
# 여기에 코드 작성
print(log.split(","))
date_text = log.split(",")[0]
code = log.split(",")[1]
status_text = log.split(",")[2]
measure_text = log.split(",")[3]
print("[" + date_text + "]")
print("[" + code + "]")
print("[" + status_text + "]")
print("[" + measure_text + "]")
# =============================================================================
#  [6] 날짜 성형 ― 슬라이싱                                          (03_02)
# -----------------------------------------------------------------------------
#  date_text("20260724")를 슬라이싱으로 쪼개
#  "2026-07-24" 형태로 만들어 date 변수에 저장하고 출력하세요.
# -----------------------------------------------------------------------------
print("[6]")
# 여기에 코드 작성
date = date_text[:4] + "-" + date_text[4:6] + "-" + date_text[6:]
print(date)


# =============================================================================
#  [7] 설비 코드 분석                                        (03_02 · 03_03 · 03_05)
# -----------------------------------------------------------------------------
#  code("PUMP-A-07")에 대해:
#
#  (1) 첫 글자를 인덱싱으로 출력                        (예상: P)
#  (2) "PUMP" 로 시작하는지 확인해 출력                 (예상: True)
#  (3) 하이픈(-) 이 몇 개인지 세어 출력                 (예상: 2)
#  (4) 하이픈 기준으로 나눠 종류/라인/번호를
#      kind, line, num 세 변수에 저장하고 출력          (예상: PUMP A 07)
# -----------------------------------------------------------------------------
print("[7]")
# 여기에 코드 작성
print(code[0])
print(code.startswith("PUMP"))
print(code.count("-"))
kind = code.split("-")[0]
line = code.split("-")[1]
num = code.split("-")[2]
print(kind, line, num)
# =============================================================================
#  [8] 상태 라벨 표준화와 비교                                       (03_04 · 03_03)
# -----------------------------------------------------------------------------
#  (1) status_text("WARNING")를 소문자로 통일해 status 에 저장하고 출력
#  (2) status 가 "warning" 과 같은지 == 로 비교해 출력   (예상: True)
#  (3) 통일하기 전 status_text == "warning" 도 출력해
#      왜 표준화가 필요한지 눈으로 확인                   (예상: False)
# -----------------------------------------------------------------------------
print("[8]")
# 여기에 코드 작성
status = status_text.lower()
print(status)
print(status == status_text)
print(status_text == "warning")

# =============================================================================
#  [9] 측정값 추출                                          (03_03 · 03_02 · 02_04)
# -----------------------------------------------------------------------------
#  measure_text("temp=91.5")에서 숫자만 뽑아냅니다.
#
#  (1) find() 로 "=" 의 위치를 찾아 출력                 (예상: 4)
#  (2) 슬라이싱으로 = 앞부분(항목 이름)을 출력            (예상: temp)
#  (3) = 뒷부분을 잘라 value_text 에 저장하고
#      값과 자료형을 출력                                (예상: 91.5 <class 'str'>)
#  (4) float() 로 변환해 value 에 저장하고
#      값과 자료형을 출력                                (예상: 91.5 <class 'float'>)
# -----------------------------------------------------------------------------
print("[9]")
# 여기에 코드 작성
position = measure_text.find("=")
print(position)
print(measure_text[:position])
value_text = measure_text[position + 1 :]
print(value_text, type(value_text))
value = float(value_text)
print(value, type(value))

# =============================================================================
#  [10] 종합 판정 ― 비교·논리 연산                                   (02_03)
# -----------------------------------------------------------------------------
#  (1) value 가 정상 범위(60 이상 90 이하)인지 → temp_ok 에 저장, 출력
#  (2) status 가 "warning" 인지 → is_warning 에 저장, 출력
#  (3) "정비 필요" 여부: 범위를 벗어났거나(or) 경고 라벨이면 True
#      → need_check 에 저장, 출력
#
#  예상 결과: False / True / True
# -----------------------------------------------------------------------------
print("[10]")
# 여기에 코드 작성
temp_ok = 90 >= value >= 60
print(temp_ok)
is_warning = status == "warning"
print(is_warning)
need_check = temp_ok == False or is_warning == True
print(need_check)

# =============================================================================
#  [11] 최종 리포트 ― f-string                                       (03_06)
# -----------------------------------------------------------------------------
#  지금까지 만든 변수들로 아래 리포트를 f-string 으로 완성하세요.
#  (구분선은 "=" * 34, "-" * 34 / 온도·가동률·불량률은 소수점 1자리)
#
#     ==================================
#     설비 점검 리포트  (2026-07-24)
#     ----------------------------------
#     설비 코드 : PUMP-A-07  (PUMP 계열 / A 라인 / 07호기)
#     상태 라벨 : warning
#     온도 측정 : 91.5 도
#     정상 범위(60~90) : False
#     가동 시간 : 28시간 45분
#     가동률 : 87.5%  /  불량률 : 3.5%
#     정비 필요 여부 : True
#     ==================================
# -----------------------------------------------------------------------------
print("[11]")
# 여기에 코드 작성
print("=" * 34)
print("설비 점검 리포트" + "(2026-07-24)" + "(03_06)")
print("-" * 34)
print(
    f"설비 코드: {equip_name}\n 상태 라벨: {status}\n 온도측정: {equip_temp}도\n 정상범위(60~90): {is_warning}\n 가동 시간: {total_min%60}시간 {total_min//60}분 \n가동률 : {run_rate}% / 불량률 : {defect_rate:.2}% \n 장비 필요 여부 : {need_check}"
)
print("=" * 34)
# =============================================================================
#  [도전] 현장 입력 버전 ― input()                                   (02_04)
# -----------------------------------------------------------------------------
#  주석을 풀고, 작업자가 직접 측정값을 입력하는 버전을 완성하세요.
#
#  · 온도(실수)와 진동(실수)을 input 으로 받고
#  · 온도 정상(60~90), 진동 정상(0.5~3.0)을 각각 판정한 뒤
#  · f-string 으로 아래처럼 출력
#
#     온도 정상: True / 진동 정상: False
#     모두 정상: False
#
#  ※ 온라인 연습장에서도 input() 이 동작합니다 (입력창이 뜸)
# -----------------------------------------------------------------------------
print("[도전]")
# temp = float(input("온도(℃): "))
# vib = float(input("진동(mm/s): "))
# 여기에 판정·출력 코드 작성
temp = float(input("온도(℃): "))
vib = float(input("진동(mm/s): "))
temp_check=90>=temp>=60
vib_check=3.0>=vib>=0.5
print(f"온도 정상: {temp_check} / 진동 정상: {vib_check} \n 모두 정상: {temp_check and vib_check}")
print()
print("종합 실습 끝 - 99_종합실습_정답.py 와 비교해보세요")
```
