```python
# .concat(['리스트로, 붙이고자하는, 데이터 프레임, 작성'])
# 세로로 이어붙이기

df_shot1 = pd.read_csv('data/13_diecasting_shot_1.csv')
df_shot2 = pd.read_csv('data/13_diecasting_shot_2.csv')

df_stacked = pd.concat([df_shot1, df_shot2]) # 1아레에 2가 붙음
print(df_stacked.shape)
```

```python
# 수치형 데이터
# pd.cut([데이터프레임], bins=[경계값], labels= [구간의 이름표, 항상 bins보다 하나 적음])

temps_band = pd.cut(df_cycle['온도'], bins=[0, 40, 50, 200], labels=['low', 'mid', 'high'])
# 경계에 걸치는 경계값은 무조건 왼쪽 구간으로 포함된다 ex -> 40 = low
print(temps_band.value_counts())
```

```python
print('max:',df_cycle['진동'].max(),'min:', df_cycle['진동'].min())
vib_band = pd.cut(df_cycle['진동'], 
                  bins=[0, 0.6, 0.7, 10],
                  labels=['약함', '보통', '강함'])

print(vib_band.value_counts())
```
