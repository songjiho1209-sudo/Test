```python
import numpy as np

data = np.loadtxt(
    'data/11_diecasting_series.csv',
    delimiter=',',
    skiprows=1,
    usecols=(0, 1, 2)
)

time = data[:, 0]
ct = data[:, 1]
pressure_cyl = data[:, 2]


idx_max_ct = np.argmax(ct)

time_max_ct = time[idx_max_ct]
max_ct = ct[idx_max_ct]

print(f'사이클타임이 가장 컸던 시점: {time_max_ct}, 사이클타임 최대값: {max_ct}')


change_ct = np.abs(np.diff(ct))
idx_max_change_ct = np.argmax(change_ct)

print(
    f'사이클타임 변화량이 가장 컸던 구간: '
    f'시점 {time[idx_max_change_ct]} -> {time[idx_max_change_ct + 1]}, '
    f'변화량: {change_ct[idx_max_change_ct]}'
)

warning = (ct >= 50) | (pressure_cyl < 150)

warning_count = warning.sum()
warning_ratio = warning_count / len(warning)

print(f'경고 개수: {warning_count}')
print(f'경고 비율: {warning_ratio}')


top3_idx = np.argsort(ct)[::-1][:3]

print('사이클타임 상위 3개:')
for idx in top3_idx:
    print(f'시점: {time[idx]}')
```
